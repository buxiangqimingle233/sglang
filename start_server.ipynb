```python
from sglang.utils import wait_for_server, print_highlight, terminate_process
import os
import subprocess

# serve model name, log, tp, dp, sp, dtype, 
log_path = "/home/wangzhao/sglang/server-debug.log"
model_path = "/datasets/zhao/Meta-Llama-3.1-8B-Instruct"

host = "0.0.0.0"
port = 30000
tp = 4
sp = 1
dtype = "bfloat16"

args = [f"--model-path {model_path}", f"--dtype {dtype}", f"--host {host}", f"--port {port}"]
args += [f"--tensor-parallel-size {tp}"]
args += ["--log-requests"]


command = "python3 -m sglang.launch_server " + " ".join(args)
command = command.replace("\\\n", " ").replace("\\", " ")

parts = command.split()


with open(log_path, 'a') as log_file:
    process = subprocess.Popen(parts, text=True, stdout=log_file, stderr=subprocess.STDOUT)

wait_for_server(f"http://localhost:{port}")
```
